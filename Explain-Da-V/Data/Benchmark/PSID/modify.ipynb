```python
import pandas as pd
data = pd.read_csv('psid.csv', keep_default_na=False, na_values=[""])
data = data.reset_index(drop=True)
data['orig_index'] = data.index
```

```python
import numpy as np
numeric_attrs = [
    col for col in data.select_dtypes(include=[np.number]).columns
]
data_norm = data.copy()
for col in numeric_attrs:
    mean = data[col].mean(skipna=True)
    std = data[col].std(skipna=True, ddof=1)
    if std == 0 or np.isnan(std):
        std = 1.0  # avoid division by zero
    data_norm[col] = (data[col] - mean) / std

data = data_norm 
```

```python
data_ver_0 = data.drop(columns=['earning_change', 'orig_index'])
data_ver_1 = data
```

```python
data_ver_0.to_csv("data_ver_0.csv", index=False)
data_ver_1.to_csv("data_ver_1.csv", index=False)
```
